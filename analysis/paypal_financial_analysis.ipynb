```python
import pandas as pd
import matplotlib.pyplot as plt

# historical data set
data = {
    "Year": [2021, 2022, 2023, 2024, 2025],
    "Revenue": [25371, 27518, 29771, 31797, 33172],
    "Operating_Income": [4262, 3837, 5028, 5325, 6065],
    "Net_Income": [4169, 2419, 4246, 4147, 5233],
    "Free_Cash_Flow": [4889, 5107, 4220, 6767, 5564],
    "Diluted_EPS": [3.52, 2.09, 3.84, 3.99, 5.41],
    "Diluted_Shares": [1186, 1158, 1107, 1039, 968]
}

df = pd.DataFrame(data)
df
```

```python
# Calculated Metrics
df["Revenue_Growth"] = df["Revenue"].pct_change()
df["Operating_Margin"] = df["Operating_Income"] / df["Revenue"]
df["Net_Margin"] = df["Net_Income"] / df["Revenue"]
df["FCF_Margin"] = df["Free_Cash_Flow"] / df["Revenue"]

df
```

```python
# Create revenue chart
plt.figure(figsize=(8,5))
plt.plot(df["Year"], df["Revenue"], marker="o")
plt.title("PayPal Revenue")
plt.xlabel("Year")
plt.ylabel("Revenue ($M)")
plt.grid(True)
plt.show()
```

```python
# Margins
plt.figure(figsize=(8,5))
plt.plot(df["Year"], df["Operating_Margin"] * 100, marker="o", label="Operating Margin")
plt.plot(df["Year"], df["Net_Margin"] * 100, marker="o", label="Net Margin")
plt.title("PayPal Profitability Trends")
plt.xlabel("Year")
plt.ylabel("Margin (%)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# FCF
plt.figure(figsize=(8,5))
plt.bar(df["Year"], df["Free_Cash_Flow"])
plt.title("PayPal Free Cash Flow")
plt.xlabel("Year")
plt.ylabel("Free Cash Flow ($M)")
plt.show()
```

```python
# Diluted share count 
plt.figure(figsize=(8,5))
plt.plot(df["Year"], df["Diluted_Shares"], marker="o")
plt.title("PayPal Diluted Share Count")
plt.xlabel("Year")
plt.ylabel("Diluted Shares (M)")
plt.grid(True)
plt.show()
```

## Key Findings

- Revenue increased steadily from 2021 through 2025, although annual growth slowed.
- Operating margins recovered and improved by 2025.
- Free cash flow remained strong but showed more volatility than revenue.
- Diluted share count declined materially, supporting EPS growth through share repurchases.
- EPS growth outpaced revenue growth over the period.

```python
# Valuation comparison chart
valuation = pd.DataFrame({
    "Method": [
        "Current Market Price",
        "DCF",
        "Forward P/E",
        "P/FCF",
        "EV/EBITDA"
    ],
    "Share_Price": [
        52.53,
        87.08,
        92.74,
        65.67,
        49.17
    ]
})

valuation
```

```python
# Draw chart
plt.figure(figsize=(9, 5))
plt.bar(valuation["Method"], valuation["Share_Price"])
plt.title("PayPal Valuation Comparison")
plt.ylabel("Implied Share Price ($)")
plt.xticks(rotation=20)
plt.show()
```

## Valuation Comparison

The valuation methods produce a wide range of implied share prices.

- DCF and Forward P/E produce the highest estimates.
- P/FCF produces a more moderate valuation.
- EV/EBITDA is closest to the reference market price.
- The dispersion shows why valuation should be viewed as a range rather than a single precise number.
